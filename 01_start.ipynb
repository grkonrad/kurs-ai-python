# Pierwsze uruchomienie środowiska

Uruchom ten notatnik w Codespaces. Wybierz kernel Python w `.venv`. Dane są sztuczne.
Pierwsza komórka sprawdza środowisko, druga przygotowuje potwierdzenie do oddania.
Nie wpisuj tu hasła, kluczy API ani danych dokumentu studenckiego.


```python
from pathlib import Path
import csv
import sys
import pandas as pd
import numpy as np
import matplotlib

print("Stanowisko działa")
print("Python:", sys.version.split()[0])
print("pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Matplotlib:", matplotlib.__version__)
with Path("lpbf_demo.csv").open(encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    columns = reader.fieldnames
    records = list(reader)
row_count = len(records)
column_count = len(columns)
print("Liczba rekordów:", row_count)
print("Liczba kolumn:", column_count)
assert row_count == 12 and column_count == 7, "Sprawdź plik danych"
print("Sprawdzenie zakończone poprawnie")
```

## Potwierdzenie

W poniższej komórce uzupełnij login GitHub oraz dwa statusy. Status Education może brzmieć „wniosek złożony, oczekuję”. Status Copilota może brzmieć „Free — działa”, „Student — działa” lub opisywać problem. To nie jest test płatnej subskrypcji.
Po wykonaniu pobierz plik `potwierdzenie_startu.txt` i zapisz notatnik.


```python
github_user = "WPISZ_LOGIN"
education_status = "WPISZ_STATUS"
copilot_status = "WPISZ_STATUS"

if github_user == "WPISZ_LOGIN" or "WPISZ_STATUS" in [education_status, copilot_status]:
    print("Uzupełnij login i statusy, a następnie uruchom tę komórkę ponownie.")
else:
    from datetime import datetime, timezone
    report = (
        f"Data UTC: {datetime.now(timezone.utc).isoformat()}\n"
        f"GitHub: {github_user}\n"
        f"Python: {sys.version.split()[0]}\n"
        f"Rekordy: {row_count}; kolumny: {column_count}\n"
        f"Education: {education_status}\n"
        f"Copilot: {copilot_status}\n"
    )
    Path("potwierdzenie_startu.txt").write_text(report, encoding="utf-8")
    print(report)
    print("Zapisano potwierdzenie_startu.txt")
```
