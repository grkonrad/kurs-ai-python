## 3. Komórka, kod i pole wyniku

Notatnik składa się z **komórek**, czyli kolejnych bloków dokumentu. Potoczne „pole notatnika” może oznaczać komórkę albo obszar wyniku — będziemy używać tych dokładniejszych nazw.

- **Komórka Markdown** zawiera tekst objaśnienia, nagłówki lub tabelę. Właśnie taką czytasz. Po dwukrotnym kliknięciu zobaczysz jej tekst do edycji. `#` na początku wiersza tworzy nagłówek, a `**tekst**` — pogrubienie. To formatowanie tekstu, nie program Python.
- **Komórka Code / Python** zawiera instrukcje dla komputera. Zwykle ma oznaczenie języka Python oraz przycisk **▶** do uruchomienia. Kolory pomagają czytać składnię, ale zależą od motywu i nie określają poprawności kodu.
- **Output / wynik** pojawia się pod uruchomioną komórką kodu: może to być napis, tabela, wykres albo komunikat błędu. Nie wpisujesz tutaj programu.

**Jak pracować:** kliknij ▶ przy komórce kodu lub kliknij wewnątrz niej i naciśnij **Shift+Enter**. Poczekaj na wynik. Ten skrót w komórce Markdown kończy edycję i wyświetla sformatowany tekst. Licznik przy komórce kodu pokazuje kolejność wykonań, a nie numer zadania; znak zajętości oznacza trwające obliczenie.

Na pierwszym przejściu uruchamiaj kod **po jednej komórce, od góry**. Opcję Run All wypróbujesz dopiero na końcu.

## 5. Z pliku do tabeli

Pytanie brzmi: **ile kosztuje cały koszyk zakupów?** Dane są fikcyjne. Każdy produkt kupujemy na sztuki.

Uruchom komórkę poniżej. Zobaczysz 5 wierszy danych i 3 kolumny. Dodatkowe liczby 0–4 po lewej to **indeks wierszy**, a nie kolumna z CSV.


```python
import pandas as pd  # Włączamy bibliotekę do pracy z tabelami.
zakupy = pd.read_csv("zakupy.csv")  # Wczytujemy plik do zmiennej zakupy.
zakupy  # Tabela w ostatnim wierszu komórki zostanie wyświetlona.
```

### Co oznaczają części kodu?

- `import pandas as pd` udostępnia gotowe narzędzia biblioteki pandas pod krótką nazwą `pd`.
- `zakupy` jest **zmienną**: nazwą, pod którą przechowujemy tabelę w pamięci. Nazwę wybrał autor programu.
- `=` oznacza **przypisanie**: oblicz prawą stronę i zapamiętaj ją pod nazwą po lewej. Nie jest pytaniem, czy dwie rzeczy są równe.
- `pd.read_csv(...)` oznacza „użyj narzędzia read_csv z biblioteki pd”.
- `"zakupy.csv"` jest tekstem z nazwą pliku. Plik leży obok notatnika w głównym folderze projektu.
- Samo `zakupy` w ostatnim wierszu komórki wyświetla tabelę w notebooku. W zwykłym skrypcie do wypisania wyniku stosuje się np. `print`.

Nie ucz się tych nazw na pamięć. Na razie wystarczy wskazać: **plik wejściowy → wczytana tabela → wynik pod komórką**.


## 6. Proste obliczenie, które potrafisz sprawdzić

Dwa mleka po 3,20 zł kosztują **6,40 zł**. Dla każdej pozycji wykonamy takie samo mnożenie. Dopiszemy do tabeli w pamięci kolumnę `koszt_zl`; plik źródłowy `zakupy.csv` pozostanie bez zmian.


```python
zakupy["koszt_zl"] = zakupy["liczba_sztuk"] * zakupy["cena_za_sztuke_zl"]
suma = zakupy["koszt_zl"].sum()
print("Suma zakupów:", round(suma, 2), "zł")
zakupy
```

### Czytamy wynik i zapis obliczenia

`zakupy["liczba_sztuk"]` wybiera z tabeli kolumnę o wskazanej nazwie. Nawiasy kwadratowe `[...]` służą tu do wskazania kolumny, a `*` oznacza mnożenie. pandas mnoży wartości odpowiadających sobie wierszy. `.sum()` dodaje koszty wszystkich pozycji. `round(suma, 2)` zaokrągla liczbę do dwóch miejsc po przecinku; wydruk `29.6` oznacza tu 29,60 zł.

**Sprawdzenie:** w wierszu Mleko odczytaj **6.4**, w wierszu Jogurt **8.4**, a w podsumowaniu **29.6 zł**. Tabela ma teraz cztery kolumny. Otwórz ponownie CSV: w pliku nadal powinny być trzy.

Nie myl dwóch pytań: „Ile kosztuje jedna sztuka?” i „Ile kosztują wszystkie sztuki tego produktu w koszyku?”. Za chwilę narysujemy wykres odpowiadający na drugie pytanie.


## 7. Wykres w notatniku

Uruchom gotową komórkę. Oś pozioma pokaże produkty, oś pionowa **koszt całej pozycji w złotych**. Oczekuj pięciu słupków; najwyższy powinien dotyczyć jogurtów (4 × 2,10 zł = 8,40 zł).


```python
import matplotlib.pyplot as plt  # Biblioteka do rysowania wykresów.

fig, ax = plt.subplots(figsize=(7, 4))  # Tworzymy obraz i miejsce na wykres.
ax.bar(zakupy["produkt"], zakupy["koszt_zl"], color="#287A9B")  # Słupki.
ax.set_title("Ile kosztują pozycje w naszym koszyku?")  # Tytuł.
ax.set_xlabel("Produkt")  # Podpis osi poziomej.
ax.set_ylabel("Koszt pozycji [zł]")  # Podpis osi pionowej z jednostką.
ax.set_ylim(0, 10)  # Zakres osi kosztu: od 0 do 10 zł.
fig.tight_layout()  # Dopasowanie odstępów, aby podpisy się mieściły.
fig.savefig("wykres_zakupow.png", dpi=150)  # Zapis obrazu do pliku.
plt.show()  # Wyświetlenie wykresu pod komórką.
```

### Co tu zobaczyliśmy?

Ten sam notebook pokazał już **tekst, tabelę i obraz**. Komentarze obok instrukcji wyjaśniają ich role. `fig` przechowuje obiekt całego obrazu, a `ax` — obiekt wykresu z osiami. Nie musisz jeszcze samodzielnie pisać takich instrukcji.

W Eksploratorze odszukaj nowy plik `wykres_zakupow.png` i otwórz go. Zapis obrazu wykonuje `savefig`, a wyświetlenie go pod komórką — `show`. To dwa osobne działania.

## 8. W jaki sposób będziemy prosić agenta o kod?

**To zapowiedź zajęć nr 2, nie zadanie do wykonania z Copilotem dzisiaj.** Nie zakładamy, że umiesz napisać program. Twoim zadaniem będzie precyzyjnie powiedzieć, czego potrzebujesz, a potem sprawdzić wynik.

Opis zadania nazywamy **specyfikacją**. Dobra specyfikacja mówi: z jakich danych korzystać, co obliczyć, co pokazać, czego nie zmieniać i jak sprawdzić poprawność.

Możesz przekazać ją na trzy sposoby:

1. **W czacie:** otworzyć Chat, wybrać tryb Agent, dołączyć plik jako kontekst i wkleić polecenie. Agent dostaje opis i właściwy plik, a nie tylko jego nazwę bez zawartości.
2. **W komórce Markdown:** opisać zadanie zwykłym tekstem, zapisać notebook, dołączyć go do czatu i poprosić: „Zrealizuj specyfikację z komórki zatytułowanej Specyfikacja. Dodaj rozwiązanie w komórce kodu poniżej; zachowaj mój opis”. Samo wpisanie opisu do Markdown **nie uruchamia agenta**.
3. **W komentarzach skryptu `.py` lub komórki kodu:** zapisać opis w wierszach zaczynających się od `#`, wskazać ten plik lub komórkę agentowi i wyraźnie poprosić o implementację. Komentarze **nie wykonują obliczeń**. Przykładowy opis jest już na początku `zakupy.py`.

Po otrzymaniu rozwiązania będziemy czytać objaśnienie, oglądać zmienione pliki, uruchamiać kod i porównywać wynik z prostym rachunkiem. Sam napis agenta „gotowe” nie potwierdza poprawności.

## 9. Zapisujemy wynik testu środowiska

Poniższy kod zapisze techniczne potwierdzenie w pliku tekstowym. Nie sprawdza Twojego konta ani statusu Education i Copilota. Te informacje dopiszesz **tekstem** w ostatniej komórce Markdown.


```python
from pathlib import Path  # Narzędzie do zapisu pliku.

potwierdzenie = (
    "Zajęcia nr 1 — test środowiska\n"
    "Dane: zakupy.csv\n"
    "Liczba pozycji: " + str(len(zakupy)) + "\n"
    "Suma zakupów: " + str(round(suma, 2)) + " zł\n"
    "Wygenerowany obraz: wykres_zakupow.png\n"
)
Path("potwierdzenie_startu.txt").write_text(potwierdzenie, encoding="utf-8")
print(potwierdzenie)
```

### Co robi zapis?

`len(zakupy)` liczy wiersze tabeli; `str(...)` zamienia wartość na tekst. `\n` w napisie oznacza przejście do nowego wiersza. `Path(...)` wskazuje plik, a `.write_text(...)` zapisuje w nim tekst. UTF-8 to sposób zapisu znaków, także polskich liter. Ponowne uruchomienie tej komórki zastąpi treść tego potwierdzenia bieżącym wynikiem.

### Dlaczego kolejność komórek ma znaczenie?

Komórka z wykresem korzysta z tabeli utworzonej wcześniej. Jeśli uruchomisz ją jako pierwszą w nowej sesji, Python może zgłosić **NameError** — nie zna jeszcze zmiennej `zakupy`. To nie znaczy, że plik został usunięty.

**Restart Kernel / Uruchom jądro ponownie** usuwa robocze zmienne z pamięci, ale nie usuwa zapisanych plików. Stary wykres widoczny w notebooku może pochodzić z poprzedniego uruchomienia. Wiarygodny test polega na ponownym wykonaniu kodu.

Teraz wybierz **Restart Kernel**, potwierdź restart, a następnie **Run All**. Poczekaj na zakończenie. Sprawdź ponownie sumę 29,60 zł i wykres. Komórki Markdown z Twoimi wpisami pozostaną zachowane. Nasz kod można powtarzać; nadpisze tylko dwa pliki wynikowe: obraz i potwierdzenie.
