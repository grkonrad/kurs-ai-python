## Spotkanie A — ćwiczenie 1
    **Cel:** zmienne, typy, działania i jednostki.

    Dane: P = 200 W, v = 800 mm/s, H = 100 µm, t = 40 µm. H i t trzeba przeliczyć na mm.
    Używamy $E=P/(vHt)$ w J/mm³ oraz $Q=vHt$ w mm³/s. Q jest nominalną szybkością przetwarzania objętości, bez czasu nakładania warstw i przestojów. E nie jest modelem porowatości.

    `=` przypisuje wartość. `print()` wyświetla wynik. `type()` pokazuje typ. `int` to liczba całkowita, `float` — zmiennoprzecinkowa, `str` — tekst, a `bool` — wartość logiczna.


```python
P_W = 200
v_mm_s = 800
h_um = 100
t_um = 40
h_mm = h_um / 1000
t_mm = t_um / 1000
E_J_mm3 = P_W / (v_mm_s * h_mm * t_mm)
Q_mm3_s = v_mm_s * h_mm * t_mm
print(type(P_W), type(h_mm))
print("E =", E_J_mm3, "J/mm³")
print("Q =", Q_mm3_s, "mm³/s")
```

```python
def nominal_rate(v_mm_s, h_mm, t_mm):
    return v_mm_s * h_mm * t_mm

print(nominal_rate(800, 0.10, 0.04), "mm³/s")
```
